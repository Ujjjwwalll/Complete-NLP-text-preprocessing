## Stemming 

### Using porter stemmer

```python
# sample words to compare the stemmers
words = ['eating','eats','eaten','writing','writes','programming','programs','history','finally','finallized']
```

```python
# printing each word with its stemmed form
from nltk.stem import PorterStemmer
porter_stemmer = PorterStemmer()
for word in words:
    print(word+" ----> "+porter_stemmer.stem(word))
```

One major disadvantage of stemming is that sometimes some words can be changed in to another where its meaning makes no sense , `for example = histori`.... This can be fixed using lemmetization

```python
# one more wrong word example 
print(porter_stemmer.stem('Congratulations'))
```

### Using Regular Expression stemmer class

```python
# removes ing / s / e / able from the end, min=4 means words shorter than 4 letters are left as it is
from nltk.stem import RegexpStemmer
regex_stem = RegexpStemmer('ing$|s$|e$|able$',min=4)
```

```python
print(regex_stem.stem('eating'))
print(regex_stem.stem('actionable'))
print(regex_stem.stem('ingeating')) # will remove ing in last only as $ sign is in last 
```

### Using Snowball stemmer 

better than porter stemmer in case of accuracy 

```python
# snowball is like an improved version of porter
from nltk.stem import SnowballStemmer
Snowball = SnowballStemmer('english')
```

```python
# porter stemmer performing wrong
print(porter_stemmer.stem('fairly'))
print(porter_stemmer.stem('sportingly'))
print(porter_stemmer.stem('invade')) # still wrong
```

```python
# now using snowball stemmer on the same
print(Snowball.stem('fairly'))
print(Snowball.stem('sportingly'))
```

## Lemmatization 

Lemmatization is the process of reducing a word to its valid dictionary root form (lemma), which directly fixes stemming's biggest drawback of producing meaningless, chopped-off non-words.

```python
# wordnet lemmatizer needs wordnet data (nltk.download('wordnet'))
from nltk.stem import WordNetLemmatizer
```

```python
# making the lemmatizer object
lemmatizer = WordNetLemmatizer()
```

```python
lemmatizer.lemmatize('going') # but there is two parameters
```

```python
lemmatizer.lemmatize('going',pos='v') # by default the value of pos is noun 'n' and we can change it accordingly and the lemmatization of words also depends on this directly.
```

```python
# lemmatizing all words as verbs
words = ['eating','eats','eaten','writing','writes','programming','programs','history','finally','finallized','fairly','sportingly']
for word in words:
    print(word+" ---> "+lemmatizer.lemmatize(word,pos='v'))
```

but there is one drawback of `lemmatization` is that it is `slower as compared to stemming` as it compares the words against a dictionary.

The main difference is that `stemming crudely chops off word endings` using fast, rigid rules, while `lemmatization uses a dictionary and grammar context` to return a real, valid base word
