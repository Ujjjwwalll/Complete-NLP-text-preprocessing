### Word2Vec

```python
# gensim for word2vec, nltk for tokenizing
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
```

```python
# same corpus again, since this is a separate notebook now
corpus = [
    "I love machine learning",
    "I love deep learning",
    "Machine learning is interesting",
    "I love machine learning and machine learning"
]

# Lowercase and tokenize
tokenized_corpus = [ word_tokenize(sentence.lower()) for sentence in corpus ]

print(tokenized_corpus)
```

Train the Word2Vec model

```python
# vector_size = length of each word vector, window = how many neighbour words to look at
# min_count=1 keeps every word since our corpus is tiny
model = Word2Vec(
    sentences=tokenized_corpus,
    vector_size=100,
    window=2,
    min_count=1,
    sg=1, # skipgram , if sg = 0 then cbow
    workers=4,
    epochs=100
)
```

```python
# getting the vector for the word 'machine'
vector = model.wv["machine"]
```

```python
# printing the vector and its shape (should be 100 values)
print(vector)
print(vector.shape)
```

```python
# Find similar words
print(model.wv.most_similar("machine", topn=3))
```
