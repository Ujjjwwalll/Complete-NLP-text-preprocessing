## Tokenization 

```python
# word_tokenize splits text into words and punctuations
import nltk
from nltk.tokenize import word_tokenize

text = "Ujjwal is learning Machine Learning in India."
tokens = word_tokenize(text)
print(tokens)
```

```python
# another example sentence
from nltk.tokenize import word_tokenize
text_2 = "I want to get a job in AI/ML. And i am a frsher."
tokenized_word = word_tokenize(text_2)
print(tokenized_word)
# Every word is a token 
```

```python
# here one sentence is one token 
from nltk.tokenize import sent_tokenize
sentence_token = sent_tokenize(text_2)
print(sentence_token)
```

```python
# treats punctuations also as a separate tokens
from nltk.tokenize import wordpunct_tokenize
punc_tokenize = wordpunct_tokenize(text_2)
print(punc_tokenize)
```

```python
# in this punctuations which are present in between the sentences will not count as separate tokens but those are present at last will be counted.from nltk.tokenize import TreebankWordTokenizer
tokenizer = TreebankWordTokenizer()
print(tokenizer.tokenize(text_2))
```

```python
# spacy gives token, pos tag and lemma all in one go
# (install model first: python -m spacy download en_core_web_sm)
# using Spacy library
import spacy

nlp = spacy.load("en_core_web_sm")

text = "Ujjwal is learning Machine Learning in India."

doc = nlp(text)

for token in doc:
    print(token.text, token.pos_, token.lemma_)
```
