## Text vectorization V/S Embedding

### `Text vectorization` is the broad process of turning any text into numbers, while `Embedding` is a smart type of text vectorization that captures the actual meaning of words.

Bag of words - text vectorization technique

```python
# small corpus of 4 sentences, using the same for bow and tfidf
corpus = [
    "I love machine learning",
    "I love deep learning",
    "Machine learning is interesting",
    "I love machine learning and machine learning"
]
```

```python
# In real-world NLP, we generally use CountVectorizer from Scikit-learn
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(stop_words='english')
```

```python
# fit_transform learns the vocabulary and gives count matrix (one row = one sentence)
X = cv.fit_transform(corpus)

print(cv.get_feature_names_out())
print(X.toarray())
```

### TF - IDF

```python
# tfidf = word count adjusted by how rare the word is across sentences
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(stop_words='english')
```

```python
# values are tfidf scores now, not raw counts
x = tfidf.fit_transform(corpus)

print(tfidf.get_feature_names_out())
print(x.toarray())
```
