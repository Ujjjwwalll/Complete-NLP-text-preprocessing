## POS tagging

```python
# first tokenize then tag every word with its part of speech
import nltk
from nltk.tokenize import word_tokenize

text = 'President Abdul Kalam inspired millions of students.'
words = word_tokenize(text)
print(words)

pos_tags = nltk.pos_tag(words)
print(pos_tags)
```

## Named Entity Recognition (NER)

```python
# example sentence for NER
Sentence= """ The Eiffel Tower was built from 1887 to 1889 by French engineer Gustave Eiffel 
whose company specialized in building metal frameworks and structures."""
```

```python
# tokenizing the sentence
words = nltk.word_tokenize(Sentence)
words
```

```python
# ne_chunk needs pos tags so tagging first
tag_elemnets=nltk.pos_tag(words)
```

```python
# opens the tree in a pop up window 
nltk.ne_chunk(tag_elemnets).draw()
```
